```python
modelo = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

print("Modelo Sentence Transformers cargado correctamente")
```

```python
embeddings = modelo.encode(
    df["texto"].tolist(),
    normalize_embeddings=True,
    show_progress_bar=False
)

print("Matriz de embeddings:", embeddings.shape)

assert len(embeddings) == len(df)
assert embeddings.shape[1] == 384
```

```python
from chromadb.errors import NotFoundError

client = chromadb.PersistentClient(
    path="/content/chroma_lab"
)

nombre = "politicas_empresa"

try:
    client.delete_collection(name=nombre)
    print("Colección anterior eliminada")

except NotFoundError:
    print("La colección no existe, se creará una nueva")

print("Cliente ChromaDB creado correctamente")
```

```python
coleccion = client.create_collection(
    name=nombre,
    configuration={"hnsw": {"space": "cosine"}}
)

print("Colección creada:", nombre)
```

```python
coleccion.add(
    ids=df["id"].tolist(),
    documents=df["texto"].tolist(),
    embeddings=embeddings.tolist(),
    metadatas=[
        {"categoria": c}
        for c in df["categoria"]
    ]
)

print("Documentos indexados:", coleccion.count())

assert coleccion.count() == len(df)
```

```python
def buscar(texto, k=3, categoria=None):

    v = modelo.encode(
        [texto],
        normalize_embeddings=True
    ).tolist()

    parametros = {
        "query_embeddings": v,
        "n_results": k,
        "include": [
            "documents",
            "metadatas",
            "distances"
        ]
    }

    if categoria is not None:
        parametros["where"] = {
            "categoria": categoria
        }

    r = coleccion.query(**parametros)

    return pd.DataFrame({
        "id": r["ids"][0],
        "documento": r["documents"][0],
        "categoria": [
            m["categoria"] for m in r["metadatas"][0]
        ],
        "distancia_coseno": r["distances"][0]
    })

print("Función de búsqueda semántica creada correctamente")
```

```python
evaluacion = [
    {
        "consulta": "¿Me devuelven dinero por un producto fallado?",
        "relevantes": {
            "DOC001", "DOC002", "DOC014",
            "DOC015", "DOC016"
        }
    },
    {
        "consulta": "¿Dónde reviso el estado de mi envío?",
        "relevantes": {"DOC005", "DOC019"}
    },
    {
        "consulta": "¿Cuándo se realiza el inventario físico?",
        "relevantes": {"DOC008"}
    },
    {
        "consulta": "¿Cómo veo los ingresos por canal?",
        "relevantes": {"DOC011", "DOC027", "DOC030"}
    },
    {
        "consulta": "¿Qué hago cuando no llega mi pedido?",
        "relevantes": {"DOC006", "DOC018", "DOC020"}
    }
]

print("Consultas de evaluación:", len(evaluacion))
```

```python
def metricas(ids_recuperados, relevantes, k):

    aciertos = len(
        set(ids_recuperados[:k]) & relevantes
    )

    precision = aciertos / k

    recall = (
        aciertos / len(relevantes)
        if relevantes else 0.0
    )

    return precision, recall
```

```python
filas = []

for caso in evaluacion:

    salida = buscar(caso["consulta"], k=3)

    ids = salida["id"].tolist()

    p, r = metricas(
        ids,
        caso["relevantes"],
        k=3
    )

    filas.append({
        "consulta": caso["consulta"],
        "ids_top3": ", ".join(ids),
        "precision@3": p,
        "recall@3": r
    })

resultados = pd.DataFrame(filas)

display(resultados)
```

```python
print(
    "Precision@3 media:",
    round(resultados["precision@3"].mean(), 3)
)

print(
    "Recall@3 medio:",
    round(resultados["recall@3"].mean(), 3)
)
```
